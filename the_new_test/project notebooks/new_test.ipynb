```python
# Import necessary libraries
import matplotlib
matplotlib.use('Agg')  # Use the appropriate backend for non-interactive plotting
import pandas as pd
from pathlib import Path
from risk_validation.core.services.pd.metrics_service import PDMetricsService
from risk_validation.core.utils.plots import plot_gini, plot_roc, plot_ks_cdf_with_maxgap
from IPython.display import Image, display

# Define the data directory and file path
DATA_DIR = Path('..') / 'data'
csv_path = DATA_DIR / 'sample.csv'

# Load the dataset
try:
    df = pd.read_csv(csv_path)
    print(f'Data loaded with shape: {df.shape}')
except FileNotFoundError as e:
    raise FileNotFoundError(f"CSV file not found at {csv_path}. Ensure the file is present.")

# Define development and validation datasets
dev_year = 2019
val_year = 2020
df_dev = df[df['score_year'] == dev_year]
df_val = df[df['score_year'] == val_year]

# Initialize the metrics service
metrics = ['auc_roc', 'gini', 'ks']
service = PDMetricsService(metrics)

# Define parameters for metric computation
params = {
    'y_col': 'default_flag',  # Actual target column
    'p_col': 'score_pd',      # Predicted probability column
    'score_col': 'score_pd'   # Score column for binning
}

# Compute metrics for development data
result_dev = service.compute(df_dev, params)
print("Development metrics:", result_dev)

# Compute metrics for validation data
result_val = service.compute(df_val, params)
print("Validation metrics:", result_val)

# Initialize directory for plots
plot_dir = Path('plots')
plot_dir.mkdir(exist_ok=True)

# Plot Gini
gini_plot_path = plot_gini(
    y_true=df_val['default_flag'],
    y_score=df_val['score_pd'],
    title='Gini Curve - Validation Dataset',
    out_path=plot_dir / 'gini_curve.png'
)
display(Image(filename=str(gini_plot_path)))

# Plot ROC Curve
roc_plot_path = plot_roc(
    y_true=df_val['default_flag'],
    y_score=df_val['score_pd'],
    title='ROC Curve - Validation Dataset',
    out_path=plot_dir / 'roc_curve.png'
)
display(Image(filename=str(roc_plot_path)))

# Plot KS CDF
ks_plot_path, _ = plot_ks_cdf_with_maxgap(
    y_true=df_val['default_flag'],
    y_score=df_val['score_pd'],
    title='KS Curve - Validation Dataset',
    out_path=plot_dir / 'ks_curve.png'
)
display(Image(filename=str(ks_plot_path)))
```
